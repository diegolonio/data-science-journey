## Basic statistics

- **Random sample**: data must be sampled randomly to avoid bias
- **Resampling**: perhaps our data is already biased, thus we have to perform a resampling for minimizing it.
- **Simple random sample**: pick rows from data randomly.
- **Stratified random sample**: our data might have multiple classes, so we want to sample it proportionaly to each of them.
- **Random sample with replacement (bootstrapping)**: take a random sample, analize it and return it to the dataset. The same sample might be taken many times, the problem is that this can increment the bias that the dataset might already has.

## Descriptive statistics

### Measures of central tendency

- **Mean**: The mean is sensitive to outliers, if just one value is abnormally higher than the rest, the mean increments drastically. $$\overline{x} = \frac{\sum_1^n{x_i}}{n}$$
- **Median**: It takes the mid value from the ordered dataset, or the mean of the mid two values if the dataset is even. The median is more robust to outliers since it stills considering the central value even if the dataset has abnormal values.
- **Mode**: The most repeated value in the dataset. It can be more than one most repeated value then we refer to it as multimodal distribution.

### Measures of spread

- **Range**: The distance between the smallest value and the largest value. It only tells to us how dispersed is the data as an overall. $$\text{Range}=\text{max(}X\text{)}-\text{min(}X\text{)}$$
- **Variance**: It describes how far apart observations are from their mean. $$s^2=\frac{\sum_1^n{(x_i-\overline{x})^2}}{n-1}$$
