Principal Component Analysis

```python
features = df.drop(columns="country")
X = features.to_numpy()
names = features.columns
countries = df["country"]
```

Standardization

```python
mu = X.mean(axis=0)
```

```python
sigma = X.std(axis=0, ddof=1)
```

```python
X_scaled = (X - mu) / sigma
```

Covariance matrix

```python
m = X_scaled.shape[0]
cov = (X_scaled.T @ X_scaled) / (m - 1)
```

```python
np.allclose(cov, np.cov(X_scaled, rowvar=False))
```

```python
vals, vecs = np.linalg.eigh(cov)
```

```python
vals
```

```python
vecs
```

```python
idx = np.argsort(vals)[::-1]
```

```python
vecs = vecs[:, idx]
```

```python
vals = vals[idx]
```

```python
per = (vals / np.sum(vals)) * 100
per
```

```python
per_acc = np.cumsum(per)
per_acc
```

```python
k = 5 # Number of eigenvectors that capture 94.5% of the info
```

```python
D = vecs[:, :k]
pd.DataFrame(D, index=names, columns=[f"PC{i+1}" for i in range(k)])
```

```python
data1 = X_scaled[0]
data1
```

```python
data1_c = data1 @ D
data1_c
```

```python
print(data1_c @ D.T)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

pipe = Pipeline([
    ("scaler", StandardScaler()),
    ("pca", PCA(n_components=5))
])

C = pipe.fit_transform(X)
pca = pipe["pca"]
```

```python
print(pca.explained_variance_ratio_.round(3))
print(pca.explained_variance_ratio_.cumsum().round(3))
```

```python
cols = [f"PC{i+1}" for i in range(pca.n_components_)]
D_sk = pd.DataFrame(pca.components_.T, index=features.columns, columns=cols)
D_sk
```

```python
C_df = pd.DataFrame(C, index=df["country"], columns=cols)
```
